Defining the path of the CFLIB data folder

```python
folder_path = "/home/arbiter/IRAF"

raw_string = r"/home/arbiter/IRAF"
```

```python
import os

list_files = os.listdir(folder_path)        # creating a list of all the file names in the folder

print(list_files)
print(len(list_files))
```

loading the sharma et al 2020 preprocessed CFLIB data to select the 850 stars we want out of all the stars in the data:

```python
cflib_data_preprocessed = np.load("cflib_data.npz", allow_pickle= True)     
```

```python
cflib_data_preprocessed["fname"]        # Checking the order of the names
```

Retrieving the 850 stars

```python
stars = []

for j in cflib_data_preprocessed["fname"]:
    for i in list_files:
        if i == j:
            stars.append(i)

stars
```

```python
len(stars)
```

Preprocessing the selected stars using preprocessing.py in the same order as done in sharma et al preprocessed CFLIB

```python
import preprocessing

flux_arr = []

for i in stars:
    for j in list_files:
        if i == j:
            wavelength, flux = preprocessing.preprocess(f"/home/arbiter/IRAF/{i}")
            flux_arr.append(flux)
            
flux_arr
```
