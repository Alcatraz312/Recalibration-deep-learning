#### Selecting out some common stars between MILES and CFLIB for comparison :



```python
cflib_path = "../preprocessing/preprocess_cflib_vyom.npy"

cflib_flux = np.load(cflib_path)
```

Getting the names of the selected 850 stars in the CFLIB library

```python
cflib_preprocessed_stars = np.load("../preprocessing/cflib_data.npz")["fname"]
```

Making a dataframe of the miles data

```python
df = pd.read_csv("miles_selected.dat")
```

```python
df.head()
```

Stripping the star names off HD in the MILES dataframe for comparison

```python
def stripping(value :str):
    if "HD " in value:
        return value.removeprefix("HD ")
    
df["Name"] = df["Name"].apply(stripping)
```

```python
df.head()
```

Finding the common stars between the selected CFLIB and MILES for comparison

```python
count = 0
for i in df["Name"]:
    for j in cflib_preprocessed_stars:
        if i == j.removesuffix(".fits"):
            print(i)
            count +=1
    
print(count)    # 213

```
