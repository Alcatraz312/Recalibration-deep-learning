Defining the Flux and wavelengths and plotting the flux vs wavelength plot

```python
with fits.open(path_249) as data_249:
    hdr = data_249[0].header    # definig the primary HDU header

    crval = hdr["CRVAL1"]      # start value of the wavelength pixel
    cdelt = hdr["CDELT1"]      # step value of the wavelength pixel 

    n_wavelength_points = len(data_249[0].data[:4394])    # total number of wavelength points 
    wavelength = crval + cdelt * np.arange(n_wavelength_points)   # defining the wavelength interval(array)
    
    flux = data_249[0].data[:4394]     # defining the flux array
    
    print("Start value of wavelength array : ",crval)
    print("Maximum wavelength value : ",max(wavelength))
    print("Step value :", {cdelt})

    # plotting
    plt.figure(figsize= (10, 6))
    plt.xlabel("Wavelength")
    plt.ylabel("Flux")
    plt.plot(wavelength,flux)
    plt.title("Spectral Continuum")
    plt.grid()
    plt.show()
```

#### Interpolator curve

```python
INTERPOLATOR_MODEL = '../RF_base_model_pvk.sav'
STANDARD_SCALER = '../rf_standard_scale_pvk.sav'
```

```python
from sklearn.externals import joblib

sc = joblib.load(STANDARD_SCALER)
ip = joblib.load(INTERPOLATOR_MODEL)
```

```python
ip.n_jobs = 1 

def InterpolateSpectrum(temperature, gravity, metallicity):
    X = np.array([temperature, gravity, metallicity]).reshape(1, -1)
    X_std = sc.transform(X)
    y = ip.predict(X_std)
    return y
```

```python
pred = InterpolateSpectrum(4757,2.88,-0.29)
```

```python
plt.figure(figsize = (12,6))
plt.plot(wavelength,pred[0,:])
plt.grid()
plt.plot()
```

#### Comparing both the curves : 

```python
plt.figure(figsize=(12,8))
plt.plot(wavelength,flux, label = "CFLIB Fits")
plt.plot(wavelength,pred[0,:], label = "RF_interpolator")
plt.title("HD 249")
plt.grid()
plt.legend()
```

Plotting the residual

```python
plt.plot(wavelength,np.abs(flux - pred[0,:]))
```
